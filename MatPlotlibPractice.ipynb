```python
import numpy as np
import matplotlib.pyplot as plt
#data
a=np.linspace(0,10,20)
b=np.linspace(10,30,20)
#graph
fig=plt.figure(figsize=(4,2))
ax1=plt.subplot()
print(a)
print(b)
ax1.plot(a)
ax1.plot(b)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
#data
a=np.linspace(0,10,20)
b=np.linspace(5,0,20)
#graph
fig=plt.figure(figsize=(5,6))
print(a)
print(b)
ax1=plt.subplot(221)
ax2=plt.subplot(224)
ax1.plot(a,linewidth=1.0,linestyle='-',color='y',alpha=1,marker='*')
ax2.plot(a,linewidth=1.0,linestyle='--',color='r',alpha=0.4,marker='^')
ax1.set(title='a - title')
ax2.set(title='b - title')
#plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
a=range(0,5)
b=[[10,2,35,4,5],[42,5,10,25,20],[2,13,4,51,6]]
print(a)
print(b)
fig=plt.figure(figsize=(10,5))
ax1=plt.subplot()
ax1.stackplot(a,b)
plt.title('Area Plot')
plt.xlabel('A X_label')
plt.ylabel('B Y_label')
```

```python
import matplotlib.pyplot as plt
data={'Apple':10,'Orange':15,'Lemon':5,'Lime':20}
names=list(data.keys())
values=list(data.values())
fig=plt.figure(figsize=(4,2))
ax1=plt.subplot()
ax1.bar(names,values)
plt.title('Bar Plot')
plt.xlabel('Names')
plt.ylabel('Values')
ax1.bar_label(ax1.bar(names,values))
plt.show()
```

```python
import matplotlib.pyplot as plt
data={'Apple':10,'Orange':15,'Lemon':5,'Lime':20}
names=list(data.keys())
values=list(data.values())
fig=plt.figure(figsize=(4,2))
ax1=plt.subplot()
ax1.barh(names,values,color='teal')
plt.title('Bar Plot')
plt.xlabel('Names')
plt.ylabel('Values')
ax1.bar_label(ax1.barh(names,values))
plt.show()
```

```python
import matplotlib.pyplot as plt
a=[10,20,30,40,50,60,70,80]
x=[5,2,4,2,1,4,5,2]
y=[1,2,3,4,3,2,1,2]
fig=plt.figure(figsize=(5,4))
ax1=plt.subplot()
ax1.scatter(a, x, c='y',s=170,edgecolor='k',marker='o',alpha=0.9)
ax1.scatter(a,y,c='r',s=140,edgecolor='g',marker='^',alpha=1)
plt.title('Scatter Plot')
plt.grid(True)
```

```python
import matplotlib.pyplot as plt
total=[20,4,1,30,20,71,30,10]
order=[10,3,1,15,17,2,30,44,2,1]
discount=[30,20,10,5,210,10,60,20,40,1]
data=list([total,order,discount])
fig=plt.figure(figsize=(3,3))
ax1=plt.subplot()
ax1.boxplot(data,showmeans=True)
plt.title('BoxPlot')
plt.grid(True)
plt.savefig
plt.show()
```

```python
import matplotlib.pyplot as plt
total=[20,4,1,30,20,71,30,10]
order=[10,3,1,15,17,2,30,44,2,1]
discount=[30,20,10,5,210,10,60,20,40,1]
data=list([total,order,discount])
fig=plt.figure(figsize=(3,3))
ax1=plt.subplot()
ax1.violinplot(data,showmeans=True)
plt.title('ViolinPlot')
plt.grid(True)
plt.savefig
plt.show()
```

```python
import matplotlib.pyplot as plt
lab=['Frogs','Hogs','Dogs','Logs']
sizes=[15,30,45,100]
exp=(0.2,0,0.3,0)
fig=plt.figure(figsize=(5,7))
ax1=plt.subplot()
ax1.pie(sizes,labels=lab,autopct='%0.3f%%',shadow=True,explode=exp,startangle=0)
plt.title('PieChart')
```
